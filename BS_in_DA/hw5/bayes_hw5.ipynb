### Задача 2

> Главная цель задачи &mdash; визуально увидеть разницу при использовании обычного и стохастического градиентов. Поэтому все параметры оптимизации, включая начальное приближение, постарайтесь выбрать так, чтобы получить как можно более наглядные графики. Попробуйте также для себя поэкспериментировать с разными параметрами задачи.

Пусть $X = (X_1, ..., X_n)$ &mdash; выборка из экспоненциального распределения с параметром $\theta$, который имеет априорное распределение $\Gamma(\alpha_0, \beta_0)$ с плотностью $q_{\alpha_0\beta_0}(t)$. Выпишите апостериорное распределение $q(t|x)$.

$$
q(t|x) = \frac{\alpha_0^{\beta_{0}} t^{\beta_0 -1}}{\Gamma(\beta_0)}\exp(-\alpha_0 t) \cdot t^{n}\exp\left( -t \sum\limits_{i=1}^{n}x_i \right)
$$

Предположим, что апостериорное распределение нам неизвестно, и мы хотим его приблизить с помощью вариационного семейства распределений $\mathcal{Q} = \{\Gamma(\alpha, \beta)\:|\:\alpha>0, \beta>0\}$, плотность которого обозначим $q_{\alpha\beta}(t)$. Заметим, что в данном случае $\mathcal{Q}$ &mdash; сопряженное семейство.

Приближение в вариационном выводе состоит в выборе распределения из $\mathcal{Q}$, наиболее похожего на апостериорное распределение. Тем самым решается задача $KL(q_{\alpha\beta}, q(\cdot|x)) \to \min\limits_{\alpha, \beta}$, которая при этом эквивалентна задаче максимизации ELBO: $L(\alpha, \beta) \to \max\limits_{\alpha, \beta}$.

Выпишите ELBO $\mathcal{L}(\alpha, \beta)$. Для упрощения задачи будем считать известным параметр $\beta$ апостериорного распределения, а именно $\beta = \beta_0+n$.

\begin{align*}
    \mathcal{L}(\alpha, \beta) 
    &= \mathbb{E}_{q_{\alpha,\beta}(\theta)} \left[ \log p(x, \theta) - \log q_{\alpha,\beta}(\theta) \right] \\
    &= (n + \beta_0 - \beta) \big( \psi(\beta) - \log \alpha \big) 
       + \left( \alpha - \alpha_0 - \sum_{i=1}^n x_i \right) \frac{\beta}{\alpha} \\
    &\quad + \beta_0 \log \alpha_0 - \log \Gamma(\beta_0) 
       - \beta \log \alpha + \log \Gamma(\beta) \\
    &\propto \left( \alpha - \alpha_0 - \sum_{i=1}^n x_i \right) \frac{\beta}{\alpha} 
       - \beta \log \alpha + \text{const} \\
\end{align*}

Выпишите производную и натуральную производную ELBO по $\alpha$, а также их стохастические версии.

Производная по $\alpha$:
$$
\frac{\partial \mathcal{L}(\alpha, \beta)}{\partial \alpha} = \frac{\beta}{\alpha} - \left( \alpha - \alpha_0 - \sum_{i=1}^n x_i \right) \frac{\beta}{\alpha^2} - \frac{\beta}{\alpha} = \left(\alpha_0 + \sum_{i=1}^n x_i - \alpha \right) \frac{\beta}{\alpha^2}
$$

Стохастическая версия:
$$
\widehat{\nabla}_{\alpha}\mathcal{L} = \left(\alpha_0 + \frac{n}{K}\sum_{i \in I} x_i - \alpha \right) \frac{\beta}{\alpha^2}
$$

Подготовимся к подсчету натрульной производной:
$$
\log(q_{\alpha, \beta}(t)) = \beta \log(\alpha) - \log\Gamma(\beta) + (\beta - 1) \log(t) - \alpha t
$$

Первая производная по $\alpha$ равна:
$$
\frac{\partial \log(q_{\alpha, \beta}(t))}{\partial \alpha} = \frac{\beta}{\alpha} - t
$$

Информация Фишера:
$$
i(\alpha) = \mathbb{E}_{q_{\alpha, \beta}(\theta)}\left[\left( \frac{\beta}{\alpha} - t \right)^2 \right] = \mathrm{Var}_{q_{\alpha, \beta}(\theta)}(t) = \frac{\beta}{\alpha^2}
$$

Натуральная производная по $\alpha$:
$$
\frac{\partial^N \mathcal{L}(\alpha, \beta)}{\partial^N \alpha} = \alpha_0 + \sum_{i=1}^n x_i - \alpha 
$$

Стохастическая версия:
$$
\widehat{\nabla}^N_{\alpha}\mathcal{L} = \alpha_0 + \frac{n}{K}\sum_{i \in I} x_i - \alpha 
$$


Реализуйте следующие функции стохастических ELBO, ее производной и натуральной производной. Аргументы функций следующие:
* `sample` &mdash; полная выборка;
* `alpha`, `beta` &mdash; параметры вариационного семейства;
* `alpha_0`, `beta_0` &mdash; параметры априорного распределения;
* `batch` &mdash; индексы выборки, образующие батч.

*Замечание.* Индексы батча генерируются независимо по принципу выбора с возвращением. Они должны генерироваться вне данных функций, это позволяет сделать шаг по градиенту и вычислить оценку ELBO на том же батче.

```python
import numpy as np
import scipy.stats as sps
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.notebook import tqdm
import numba as nb

sns.set(font_scale=1.3, palette="Set2")
```

```python
@nb.jit(nopython = True)
def elbo(sample, alpha, beta, alpha_0, beta_0):
    n = sample.shape[0]
    S = np.sum(sample)
    return beta * (alpha - alpha_0 - S) / alpha - beta * np.log(alpha)

@nb.jit(nopython=True)
def stochastic_elbo(sample, alpha, beta, alpha_0, beta_0, batch):
    n = sample.shape[0]
    K = batch.shape[0]
    S_hat = n * np.sum(sample[batch]) / K
    return beta * (alpha - alpha_0 - S_hat) / alpha - beta * np.log(alpha)

@nb.jit(nopython = True)
def stochastic_grad_elbo(sample, alpha, beta, alpha_0, beta_0, batch):
    n = sample.shape[0]
    K = batch.shape[0]
    S_hat = n * np.sum(sample[batch]) / K
    return (alpha_0 + S_hat - alpha) * beta / alpha**2

@nb.jit(nopython = True)
def stochastic_natural_grad_elbo(sample, alpha, beta, alpha_0, beta_0, batch):
    n = sample.shape[0]
    K = batch.shape[0]
    S_hat = n * np.sum(sample[batch]) / K
    return alpha_0 + S_hat - alpha
```

Выберите некоторое значение $\theta$ в качестве истинного и сгенерируйте выборку $X = (X_1, ..., X_n)$ размера $n=10^6$.

```python
n = 1e6
theta_true = 3.0
sample = sps.expon(scale=1/theta_true).rvs(size=int(n))
```

Выберите параметры априорного распределения и посчитайте параметры апостериорного.

```python
alpha_0 = 1000
beta_0 = 5000

alpha_true_post = alpha_0 + np.sum(sample)
beta_true_post = beta_0 + n

print(f"True posterior params: alpha = {alpha_true_post}, beta = {beta_true_post}")
```

Установите параметры процесса оптимизации, например, как указано ниже. Для корректной оптимизации шаг обучения должен уменьшаться с ростом номера итерации.

```python
n_iter = 100000
batch_size = 64

def learning_rate(iteration, start=1e-2):
    return start / (1 + np.log1p(iteration))
```

Выберите начальное приближение для $\alpha$ и реализуйте градиентный спуск, используя обычный стохастический градиент. Нарисуйте на разных графиках траектории полного ELBO, стохастического ELBO и значения $\alpha$. Сошелся ли процесс к параметру $\alpha$ у апостериорного распределения?

```python
def run_sgd(sample, alpha_0, beta_0, n_iter, batch_size, learning_rate, sgd_func=stochastic_grad_elbo, tol=1e-3):
    n = sample.shape[0]
    beta = beta_0 + n
    alpha = alpha_0
    K = batch_size

    ELBO_stochastic = [stochastic_elbo(sample, alpha, beta, alpha_0, beta_0, np.arange(batch_size))]
    ELBO_true = [elbo(sample, alpha, beta, alpha_0, beta_0)]
    alpha_path = [alpha]

    batches = np.random.choice(n, size=(n_iter, batch_size), replace=True)
    for i in tqdm(range(n_iter), desc="SGD Iterations"):

        batch = batches[i, :]
        grad = sgd_func(sample, alpha, beta, alpha_0, beta_0, batch)
        lr = learning_rate(i)

        alpha = alpha + lr * grad

        alpha_path.append(alpha)
        elbo_stoch = stochastic_elbo(sample, alpha, beta, alpha_0, beta_0, batch)
        ELBO_stochastic.append(elbo_stoch)
        ELBO_true.append(elbo(sample, alpha, beta, alpha_0, beta_0))

        if abs(ELBO_true[-1] - ELBO_true[-2]) < tol:
            print("CONVERGED!")
            break

    return np.array(ELBO_true), np.array(ELBO_stochastic), np.array(alpha_path)
```

```python
ELBO_true, ELBO_stochastic_list, alpha_list = run_sgd(sample, alpha_0, beta_0, n_iter, batch_size, learning_rate, stochastic_grad_elbo)
ELBO_stochastic_list.shape
```

```python
from scipy.ndimage import gaussian_filter1d

def plot_elbo_and_alpha(ELBO_true, ELBO_stochastic, alpha_list, alpha_true_post,
                        smooth_sigma=2, title="ELBO and Alpha Trajectories"):
    plt.style.use('seaborn-v0_8-whitegrid')
    fig, axes = plt.subplots(2, 1, figsize=(8, 7), sharex=True)
    fig.suptitle(title, fontsize=14, fontweight='semibold', y=0.97)
    iters = np.arange(1, 1+len(ELBO_stochastic))
    # ---------- 1. ELBO ----------
    ax = axes[0]
    ELBO_stoch_smooth = gaussian_filter1d(ELBO_stochastic, sigma=smooth_sigma)

    ax.plot(iters, ELBO_true, label='True ELBO', color='#1f77b4', lw=2)
    ax.plot(iters, ELBO_stochastic, label='Stochastic ELBO (raw)', color='orange', alpha=0.75, lw=2)
    # ax.plot(iters, ELBO_stoch_smooth, label='Stochastic ELBO (smoothed)', color='darkorange',alpha=0.7, lw=2)

    y_min = min(np.min(ELBO_true), np.min(ELBO_stochastic))
    y_max = max(np.max(ELBO_true), np.max(ELBO_stochastic))
    ax.set_ylim(y_min - 0.05*(y_max-y_min), y_max + 0.05*(y_max-y_min))

    ax.set_ylabel('ELBO', fontsize=12)
    ax.set_title('True vs. Stochastic ELBO', fontsize=11)
    ax.legend(frameon=True, loc='best', fontsize=9)
    ax.grid(True, linestyle='--', alpha=0.4)
    ax.margins(x=0)

    # ---------- 2. Alpha trajectory ----------
    ax = axes[1]
    ax.plot(iters, alpha_list, color='#2ca02c', lw=2, label=r'$\alpha$ (estimated)')
    ax.axhline(alpha_true_post, color='red', lw=2, ls='--', label=r'$\alpha_{\mathrm{true\ post}}$')

    y_min = min(np.min(alpha_list), alpha_true_post)
    y_max = max(np.max(alpha_list), alpha_true_post)
    ax.set_ylim(y_min - 0.05*(y_max-y_min), y_max + 0.05*(y_max-y_min))

    ax.set_xlabel('Iteration', fontsize=12)
    ax.set_ylabel(r'$\alpha$', fontsize=12)
    ax.set_title(r'Trajectory of $\alpha$', fontsize=11)
    ax.legend(frameon=True, loc='best', fontsize=9)
    ax.grid(True, linestyle='--', alpha=0.4)
    ax.margins(x=0)

    ax.set_xscale('log')
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
```

**2.** Данные содержат $K = 10$ кластеров по количеству различных цифр. Предположим, что пиксели независимы, причем пиксель $(h, w)$ является белым в кластере $k$ с вероятностью $\theta_{khw}$. Пусть также $\pi_k$ &mdash; вероятность кластера $k$. Тогда распределение можно представить с помощью дискретной плотности в виде
$$
p_{\theta\pi}(x) = \sum_{k=1}^K \pi_k \prod_{h=1}^H \prod_{w=1}^W \theta_{khw}^{x_{hw}} \left(1-\theta_{khw}\right)^{1-x_{hw}}.
$$
Введите латентные величины так, чтобы совместная плотность $p_{\theta\pi}(x, y)$ принадлежала экспоненциальному классу распределений. 

Как и раннее в таких случаях удобно рассмотреть y в виде onehot-вектора.
Тогда совместная плотность имеет вид:
$$
p_{\theta\pi}(x, y) = \prod_{k=1}^K \left[\pi_k  \prod_{h=1}^H \prod_{w=1}^W \theta_{khw}^{x_{hw}} \left(1-\theta_{khw}\right)^{1-x_{hw}}\right]^{y_{k}}
$$

**3.**
Выведите формулы EM-алгоритма для оценки параметров $\theta, \pi$ и для ELBO. Выпишите итоговую процедуру.

**E-шаг**
Плотность вариационного распределения совпдает с той, что была для смеси распределений:

$$
q(y) = p_{\theta, \pi}(y|x) \propto p_{\theta, \pi}(y, x)  = \prod^{n}_{i=1}p_{\theta\pi}(x_i, y_i)
$$

Отсюда видно, что наше распределение факторизуется:
$$
q(y) = \prod^{n}_{i=1}q_{i}(y_i)
$$
где $q_{i}(y_i) \propto p_{\theta\pi}(x_i, y_i) = \prod_{k=1}^{K} \gamma_{ik}^{y_{ik}}$ 

где:
$$
\boxed{
\gamma_{ik} = \frac{\pi_k p_{\theta k}(x_i)}{\sum\limits_{j=1}^{K}\pi_j p_{\theta j}(x_i)}
}
$$


M-шаг

\begin{align*}
    E_{q} \log(p_{\theta \pi}(x, y)) = \sum_{k=1}^{K}\sum_{i=1}^{n}\gamma_{ik} (\log\pi_k + \log(p_{\theta k}(x_i))) \rightarrow \max_{\pi, \theta}
\end{align*}

Далее я просто возьму вывод для смесу распределений, так как это именно наш случай:

$$
\boxed{
    \pi_{k} = \frac{1}{n}\sum_{i=1}^{n}\gamma_{ik}
}
$$

Для $\theta_{khw} имеем:
\begin{align*}
\frac{\partial E_{q} \log(p_{\theta \pi}(x, y))}{\partial \theta_{khw}} &= \sum_{i=1}^{n}\gamma_{ik} \frac{\partial}{\partial \theta_{khw}}\left[ x_{ihw}\log\theta_{khw} + (1 - x_{ihw})\log(1 - \theta_{khw})\right] \\
&=  \sum_{i=1}^{n}\gamma_{ik} \left[ \frac{x_{ihw}}{\theta_{khw}} - \frac{1 - x_{ihw}}{1 - \theta_{khw}}\right] = 0
\end{align*}

Откуда:
$$
\boxed{
    \theta_{khw} = \frac{\sum_{i=1}^{n} \gamma_{ik} x_{ihw}}{\sum_{i=1}^{n} \gamma_{ik} x_{ihw} + \gamma_{ik}(1 - x_{ihw})} = \frac{\sum_{i=1}^{n} \gamma_{ik} x_{ihw}}{\sum_{i=1}^{n} \gamma_{ik}}
}
$$


**ELBO**
$$
\mathcal{L} = \sum_{i=1}^{n} \sum_{k=1}^{K} \gamma_{ik} \Big( \log \pi_k + \log p_{\theta k}(x_i) - \log \gamma_{ik} \Big)
$$

**Итоговая процедура EM**

1. Инициализировать $\pi_k^{(0)}$ и $\theta_{khw}^{(0)}$ случайно.

2. Повторять до сходимости:
   2.1. **E-шаг:** для каждого $i,k$ вычислить
   $$
   \gamma_{ik} \leftarrow \frac{\pi_k p_{\theta k}(x_i)}{\sum_{j=1}^{K} \pi_j p_{\theta j}(x_i)}
   $$
   2.2. **M-шаг** обновить параметры
   $$
   \pi_k \leftarrow \frac{1}{n} \sum_{i=1}^{n} \gamma_{ik}, \quad
   \theta_{khw} \leftarrow \frac{\sum_{i=1}^{n} \gamma_{ik} x_{ihw}}{\sum_{i=1}^{n} \gamma_{ik}}
   $$

3. Вычислить ELBO для контроля сходимости
$$
\mathcal{L} = \sum_{i=1}^{n} \sum_{k=1}^{K} \gamma_{ik} \Big( \log \pi_k + \log p_{\theta k}(x_i) - \log \gamma_{ik} \Big)
$$

Инициализация вероятности будем из распределения Дирихле с единичными параметрами, т.е. равномерное распределение на симплексе.

```python
def initialize(K, HW):
    '''
    Случайная инициализация параметров модели.
    * K -- число классов
    * HW -- количество пикселей
    
    Возвращает:
    * log_pi -- логарифмы вероятностей классов
    * theta -- вероятности белых пикселей
    '''
    
    pi = sps.dirichlet(np.ones(K)).rvs()[0]
    theta = sps.dirichlet(np.ones(K)).rvs(size=HW).T
    
    return pi, theta
```

Функция `callback`, которая на каждой итерации рисует текущую информацию. Обратите внимание, что использование данной функции может замедлять скорость обучения, используйте ее вместе с функцией `visualize_or_not`.

```python
def callback(pi, theta, elbo_trace, optim_elbo, init, it):
    '''
    Отрисовка текущего статуса обучения.
    * pi -- оценки вероятностей кластеров на текущей итерации
    * theta -- вероятности белых пикселей на текущей итерации
    * elbo_trace -- зависимость логарифма вариационной нижней оценки от номера итерации 
        до текущей итерации включительно
    * optim_elbo -- оптимальная ELBO за все время обучения
    * init -- номер инициализации параметров
    * it -- номер итерации обучения
    
    Обратите внимание, что использование данной функции может замедлять скорость обучения.
    '''
    
    # Отложенная очистка экрана
    clear_output(wait=True)
    
    # Количество классов и количество строк
    K, HW = theta.shape
    H = int(np.sqrt(HW))
    nrows = int(np.ceil(K/5))
    
    # Объявление фигуры и сетки
    fig = plt.figure(figsize=(2*5, 2*(nrows+2)), constrained_layout=True)
    plt.suptitle(f'Инициализация {init}, Итерация {it}, оптим. ELBO = {np.round(optim_elbo)}')
    gs = GridSpec(nrows+2, 5, figure=fig)
    
    # График ELBO
    ax_elbo = fig.add_subplot(gs[:2, :3])
    ax_elbo.plot(elbo_trace, lw=2, color='#00CC66')
    ax_elbo.set_xlabel('Номер итерации')
    ax_elbo.set_title('ELBO')
    
    # График вероятностей кластеров
    with sns.axes_style("whitegrid"):
        ax_pi = fig.add_subplot(gs[:2, 3:])
        ax_pi.bar(np.arange(K), pi, color='#00CC66')
        ax_pi.set_xlabel('Номер кластера')
        ax_pi.set_title('Вероятности кластеров') 
        ax_pi.set_xticks(np.linspace(0, K-1, min(10, K)).astype(int))
    
    # График центров кластеров
    for i in range(nrows):
        for j in range(5):
            ind = 5*i + j
            if ind >= K: break
                
            ax = fig.add_subplot(gs[i+2, j])
            ax.axis('off')
            ax.imshow(theta[ind].reshape(H, H), cmap='gray')
            
    plt.tight_layout()
    plt.show()


def visualize_or_not(it, n_iter):
    '''
    Функция, которая определяет, визуализировать ли результат данной итерации.
    
    * it -- номер текущей итерации
    * n_iter -- количество итераций чередования E и M шагов
    '''
    
    return (it < 10) or ((it % 10 == 0) and (it < 100)) \
            or ((it % 100 == 0) and (it < 1000)) or (it % 1000 == 0) or (it == n_iter-1)
```

```python
class NaiveELBO:
    def __init__(self, X, Y):
        self.X = X
        self.Y = Y
        self.n, self.HW = X.shape
        self.K = len(np.unique(Y))

        self.pi, self.theta = initialize(
            self.K, self.HW
        )  # pi : K x 1, theta : K x HW
        self.elbo_trace = []
        self.gamma = np.zeros((self.n, self.K))

    def E_step(self):
        """
        E-шаг алгоритма EM.
        Вычисляет матрицу гамма -- вероятностей принадлежности объектов к классам.
        """
        eps = 1e-10
        X_flat = self.X.reshape(self.n, -1)  # n x HW
        theta_flat = self.theta.reshape(self.K, -1)  # K x HW

        log_pi = np.log(self.pi + eps)  # K x 1
        log_theta = np.log(theta_flat + eps)  # K x HW
        log_1_minus_theta = np.log(1 - theta_flat + eps)  # K x HW

        log_joint = (
            X_flat @ log_theta.T + (1 - X_flat) @ log_1_minus_theta.T + log_pi
        )  # n x K
        log_marginal = logsumexp(log_joint, axis=1, keepdims=True)  # n x 1
        log_gamma = log_joint - log_marginal  # n x K
        self.gamma = np.exp(log_gamma)  # n x K

        assert np.all(self.gamma >= 0)
        assert np.allclose(self.gamma.sum(axis=1), 1, atol=1e-6)

    def M_step(self):
        """
        M-шаг алгоритма EM.
        Обновляет параметры модели -- вероятности классов и вероятности белых пикселей.
        """
        eps = 1e-10
        X_flat = self.X.reshape(self.n, -1)  # (n, D)

        self.pi = np.mean(self.gamma, axis=0)

        numerator = self.gamma.T @ X_flat  # (K, n) @ (n, D) = (K, D)
        denominator = self.gamma.sum(axis=0)  # (K,)

        self.theta = (numerator) / (denominator[:, np.newaxis] + eps)

    def elbo(self):
        """
        Вычисляет значение вариационной нижней оценки (ELBO).
        """
        eps = 1e-10
        X_flat = self.X.reshape(self.n, -1)  # n x HW
        theta_flat = self.theta.reshape(self.K, -1)  # K x HW

        log_pi = np.log(self.pi + eps)  # K x 1
        log_theta = np.log(theta_flat + eps)  # K x HW
        log_1_minus_theta = np.log(1 - theta_flat + eps)  # K x HW

        log_p = np.sum(
            self.gamma
            * (X_flat @ log_theta.T + (1 - X_flat) @ log_1_minus_theta.T + log_pi),
        ) 
        entropy = -np.sum(self.gamma * np.log(self.gamma + eps))

        elbo_value = log_p + entropy
        self.elbo_trace.append(elbo_value)


    def fit(self, max_iter=100, tol=1e-4, verbose=True):
        """
        Запускает EM-алгоритм.

        Параметры:
        ----------
        max_iter : int, по умолчанию 100
            Максимальное число итераций.
        tol : float, по умолчанию 1e-4
            Порог изменения ELBO для остановки.
        verbose : bool, по умолчанию True
            Печатать прогресс?
        """
        self.elbo_trace = []

        prev_elbo = -np.inf
        for iteration in range(max_iter):
            self.E_step()
            self.M_step()
            self.elbo()

            current_elbo = self.elbo_trace[-1]
            delta = current_elbo - prev_elbo

            if visualize_or_not(iteration, max_iter):
                callback(self.pi, self.theta, self.elbo_trace, np.max(self.elbo_trace), 1, iteration)
                
            if verbose:
                print(f"Iteration {iteration + 1}: ELBO = {current_elbo:.4f}, Δ = {delta:.6f}")

            if delta < tol and iteration > 0:
                if verbose:
                    print("Сходимость достигнута.")
                break

            prev_elbo = current_elbo

    def predict(self):
        """
        Возвращает предсказанные метки кластеров (жёсткая классификация).

        Возвращает:
        ----------
        labels : ndarray, shape (n,)
            Предсказанные метки (от 0 до K-1).
        """
        return np.argmax(self.gamma, axis=1)
```

**7.** Модифицируйте полученную ранее процедуру для работы с большим объемом выборки, оптимизируя ELBO с помощью стохастической оценки натурального градиента. Считать сам натуральный градиент в данном случае не требуется, можно
* сгенерировать батч из $n$ элементов выборки,
* на E-шаге посчитать только те параметры, которые соответствуют выбранному батчу,
* на M-шаге и в ELBO заменить все суммы $n$ элементов на их стохастические оценки,
* для обновления параметров на M-шаге использовать шаг обучения, убывающий с увеличением номера итерации, но не слишком быстро, например, $\theta_{t+1} = \theta_{t} + \eta (\theta_{new} - \theta_{t})$, где в скобках по сути и есть натуральный градиент.

*Замечание.* Приведенная формула обновления шага должна применяться для самих параметров вероятностей, а не для их логарифмов.

**7.** Реализуйте стохастическую версию процедуры и сравните ее работу с исходной процедурой. Размер батча берите порядка 20-50 объектов. Обратите внимание, что для стохастической процедуры требуется проводить больше итераций.

```python
class StochasticELBO(NaiveELBO):
    def __init__(self, X, Y, batch_size=32):
        super().__init__(X, Y)
        self.batch_size = batch_size
        self.batches = np.random.choice(self.n, size=(100000, batch_size), replace=True)
        self.lr_function = lambda it: 0.1 / (1 + np.log1p(it))

        self.gamma_batch = None
    def E_step(self):
        """
        E-шаг алгоритма EM.
        Вычисляет матрицу гамма -- вероятностей принадлежности объектов к классам.
        """
        eps = 1e-10
        batch = self.batches[len(self.elbo_trace)]
        X_flat = self.X[batch].reshape(len(batch), -1)  # n x HW
        theta_flat = self.theta.reshape(self.K, -1)  # K x HW

        log_pi = np.log(self.pi + eps)  # K x 1
        log_theta = np.log(theta_flat + eps)  # K x HW
        log_1_minus_theta = np.log(1 - theta_flat + eps)  # K x HW

        log_joint = (
            X_flat @ log_theta.T + (1 - X_flat) @ log_1_minus_theta.T + log_pi
        )  # n x K
        log_marginal = logsumexp(log_joint, axis=1, keepdims=True)  # n x 1
        self.gamma_batch = np.exp(log_joint - log_marginal)  # (B, K)

    def M_step(self):
        """
        M-шаг алгоритма EM.
        Обновляет параметры модели -- вероятности классов и вероятности белых пикселей.
        """
        eps = 1e-10
        batch = self.batches[len(self.elbo_trace)]
        X_flat = self.X[batch].reshape(len(batch), -1)

        new_pi = np.mean(self.gamma_batch, axis=0)
        lr = self.lr_function(len(self.elbo_trace))
        self.pi = (1 - lr) * self.pi + lr * new_pi  

        numerator = self.gamma_batch.T @ X_flat
        denominator = self.gamma_batch.sum(axis=0)
        theta_new = (numerator) / (denominator[:, np.newaxis] + eps)
        self.theta = (1 - lr) * self.theta + lr * theta_new

    def elbo(self):
        """
        Вычисляет значение вариационной нижней оценки (ELBO).
        """
        eps = 1e-10
        batch = self.batches[len(self.elbo_trace)]
        X_flat = self.X[batch].reshape(len(batch), -1)
        theta_flat = self.theta.reshape(self.K, -1)

        log_pi = np.log(self.pi + eps)
        log_theta = np.log(theta_flat + eps)
        log_1_minus_theta = np.log(1 - theta_flat + eps)

        log_p = np.sum(
            self.gamma_batch
            * (X_flat @ log_theta.T + (1 - X_flat) @ log_1_minus_theta.T + log_pi),
        ) 
        entropy = -np.sum(self.gamma_batch * np.log(self.gamma_batch + eps))

        elbo_value = log_p + entropy
        self.elbo_trace.append(elbo_value)

    def fit(self, max_iter=10000, verbose=True):
        self.iteration = 0
        for it in range(max_iter):
            self.E_step()
            self.M_step()
            self.elbo()
            self.iteration += 1

            if visualize_or_not(it, max_iter):
                callback(self.pi, self.theta, self.elbo_trace, np.max(self.elbo_trace) if self.elbo_trace else -np.inf, 1, it)
```

```python
stochastic_elbo = StochasticELBO(X, y, batch_size=32)
stochastic_elbo.fit(max_iter=1024, verbose=True)
```

**8.** Определите явным образом вариационное семейство распределений, которое соответствует рассматриваемому нами EM-алгоритму, и выпишите плотность вариационного распределения, а также ELBO.

Распишите процедуру стохастического вариационного вывода для данной задачи, используя обычный градиент ELBO и двойную стохастичность &mdash; по элементам выборки и при использовании метода Монте-Карло. Укажите, каким образом формируются батчи.

*Замечание.* В данном случае можно пойти двумя способами
1. как при сведении VI к EM, считать все параметры случайными, но для некоторых из них использовать вырожденные распределения;
2. оставив параметры, которые обновляются на M-шаге, неслучайными, сведя к задаче 1.

Эти способы эквивалентны. Почему?

Заметим, что вариационным семейством является экспоненциальное семейство распределений.

Действительно:

$$
q_{i}(y_i) = \frac{g(y_i)}{h(t_i)}\exp \left(t_i^T u(y_i) \right)
$$

где $g(y_i) = 1, h(t_i)=\sum_{k=1}^{K}\exp{t_{ik}}, u(y_i) = y_i$, причем:
$$
\gamma_{ik} = \frac{\exp(t_{ik})}{\sum_{j=1}^K \exp(t_{ik})}
$$

Для этого распределения ELBO уже известен:
\begin{align*}
\mathcal{L}(\pi, \theta) &= \sum_{i=1}^{n} \sum_{k=1}^{K} \gamma_{ik} \Big( \log \pi_k + \log p_{\theta k}(x_i) - \log \gamma_{ik} \Big) \\
&= \sum_{i=1}^{n} \sum_{k=1}^{K} \gamma_{ik} \Big( \log \pi_k + \log \left[\prod_{h=1}^H \prod_{w=1}^W \theta_{khw}^{x_{hw}} \left(1-\theta_{khw}\right)^{1-x_{hw}}\right] - \log \gamma_{ik} \Big) \\
&= \sum_{i=1}^{n} \sum_{k=1}^{K} \gamma_{ik} \Big( \log \pi_k + \sum_{h=1}^H \sum_{w=1}^W x_{hw}\log(\theta_{khw}) +(1-x_{hw})\log\left(1-\theta_{khw}\right)- \log \gamma_{ik} \Big)
\end{align*}

Займемся стохастической версией ELBO

Пусть I = {i_1,..., i_L} \sim U(1, ..., n)

Тогда стохастический ELBO имеет вид:
$$\boxed{
\hat{\mathcal{L}}(\pi, \theta) = \frac{n}{L}\sum_{i \in I} \sum_{k=1}^{K} y_{ik} \Big( \log \pi_k + \sum_{h=1}^H \sum_{w=1}^W x_{hw}\log(\theta_{khw}) +(1-x_{hw})\log\left(1-\theta_{khw}\right)- \log \gamma_{ik} \Big)
}
$$

Перед тем как записать формулу обновления параметров, ответим на вопрос, почему указанные способы эквивалентны.

В первом случае мы берем вырожденные распределения для $\theta$ и $\pi$, и при взятии мат ожидания в ELBO, мы получим наш M-шаг, так как логарифмах будут текущие значения $\theta$ и $\pi$.

Во втором случае, мы сразу же утверждаем, что $\theta$ и $\pi$ это параметры, и получим наш M-шаг, где происходит их оптимизация.

Теперь, когда мы показали эквивалентность, выпишем стохастическое обновление параметров как в исходной задаче:

$$
\boxed{
    \pi_k \leftarrow (1 - \eta_t)\,\pi_k + \eta_t \cdot \frac{1}{|I|} \sum_{i \in I} \gamma_{ik}
}
$$

$$
\boxed{
    \theta_{khw} \leftarrow (1 - \eta_t)\,\theta_{khw} + \eta_t \cdot \frac{\sum_{i \in I} \gamma_{ik} x_{ihw}}{\sum_{i \in I} \gamma_{ik}}
}
$$

**9.** Реализуйте процедуру, полученную в предыдущем пункте. Учтите также замечания и указания в предыдущих пунктах. В данном случае может потребоваться еще больше итераций.

```python
class StochasticMCELBO(StochasticELBO):
    def __init__(self, X, Y, batch_size=32):
        super().__init__(X, Y, batch_size)
    def elbo(self):
        """
        Вычисляет значение вариационной нижней оценки (ELBO).
        """
        eps = 1e-10
        batch = self.batches[len(self.elbo_trace)]
        y_samples = np.zeros_like(self.gamma_batch)
        for i in range(len(self.gamma_batch)):
            k = np.random.choice(self.K, p=self.gamma_batch[i])
            y_samples[i, k] = 1

        X_flat = self.X[batch].reshape(len(batch), -1)
        theta_flat = self.theta.reshape(self.K, -1)

        log_pi = np.log(self.pi + eps)
        log_theta = np.log(theta_flat + eps)
        log_1_minus_theta = np.log(1 - theta_flat + eps)

        log_p = np.sum(
            y_samples * (X_flat @ log_theta.T + (1 - X_flat) @ log_1_minus_theta.T + log_pi)
        )
        entropy = -np.sum(y_samples * np.log(self.gamma_batch + eps)) 

        elbo_value = log_p + entropy
        self.elbo_trace.append(elbo_value)
```

```python
stochastic_mc_elbo = StochasticMCELBO(X, y, batch_size=32)
stochastic_mc_elbo.fit(max_iter=1024, verbose=False)
```
