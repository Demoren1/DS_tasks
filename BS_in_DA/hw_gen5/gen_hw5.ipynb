**1.** Рассмотрим в качестве целевого распределения $\mathsf{P}_0$ многомерное нормальное $\mathcal{N}(a, \Sigma)$. Найдите распределение зашумленного объекта на шаге $t$, используя явную формулу, а также производную его плотности по аргументу.

Мы уже вывели нужные формулы в прошлых номерах, просто применим их.

$X_t = \sqrt{\tilde{\alpha}_t} X_0 + \sqrt{(1 - \tilde{\alpha}_t)}\varepsilon_t \sim N(\sqrt{\tilde{\alpha}_t}a,\tilde{\alpha}_t\Sigma + (1 - \tilde{\alpha}_t)I_d)$

Обозначим дисперсию за A.

Тогда имеем следующую плотность

$$
p_t(x) = (2\pi \det A)^{-d/2}\exp\left(-\frac{1}{2}(x - \sqrt{\tilde{\alpha}_t}a)^T A^{-1}(x - \sqrt{\tilde{\alpha}_t}a)\right)
$$

Продифференцируем, получим:

$$
\nabla_{x}p_{t}(x) = -p_{t}(x)A^{-1}(x - \sqrt{\tilde{\alpha_t}}a)
$$

Заодно получим сразу градиент логарифма, он понадобится нам далее:
$$
\nabla_{x}\log(p_{t}(x)) = -A^{-1}(x - \sqrt{\tilde{\alpha_t}}a)
$$

Выпишите дифференциальные уравнения:
* прямое стохастическое,
* обратное стохастическое,
* обратное обыкновенное.

Укажите также начальное распределение.

Прямое стохастическое:

$$
dX_t = -\frac{1}{2}\beta_t X_t dt + \sqrt{\beta_t}dW_t,
$$
где $W_t$ -- итерация винеровского процесса.

Обратное стохастическое:
$$
dX_t = \left(-\frac{1}{2}\beta_t X_t +\beta_t A^{-1}(X_t - \sqrt{\tilde{\alpha_t}}a) \right)dt + \sqrt{\beta_t}dW_t
$$

Обратное обыкновенное:
$$
dX_t = \left(-\frac{1}{2}\beta_t X_t + \frac{1}{2}\beta_t A^{-1}(X_t - \sqrt{\tilde{\alpha_t}}a)\right)dt
$$

Начальное распределение:
$$
X_0 \sim N(a, \Sigma)
$$
$$
X_T \sim N(0, I_d)
$$

```python
import numpy as np
import scipy.stats as sps
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
a = np.array([2.0, 2.0])
Sigma = np.array([[2.0, 0.5], [0.5, 1.0]])
Sigma_inv = np.linalg.inv(Sigma)

P_0 = sps.multivariate_normal(mean=a, cov=Sigma)
```

```python
T = 1000
t_s = np.arange(0, T)
beta_t = np.linspace(1e-2, 1e-4, T)
alpha_t = 1 - beta_t
alpha_hat_t = np.cumprod(alpha_t)
```

```python
n_dots = 100
d = 2

means_t = np.sqrt(alpha_hat_t)[:, None] * a
covs_t = (
    alpha_hat_t[:, None, None] * Sigma[None]
    + (1 - alpha_hat_t)[:, None, None] * np.eye(d)
)

X_t = np.zeros((t_s.shape[0], n_dots, d))

for i in range(t_s.shape[0]):
    X_t[i] = sps.multivariate_normal(
        mean=means_t[i],
        cov=covs_t[i],
    ).rvs(size=n_dots)
```

```python
x = np.linspace(-4, 5, 500)
y = np.linspace(-3, 5, 500)
Xg, Yg = np.meshgrid(x, y)
pos = np.dstack((Xg, Yg))

Z_p0 = P_0.pdf(pos)
Z_std = sps.multivariate_normal(mean=[0, 0], cov=np.eye(2)).pdf(pos)

sns.set_style("whitegrid")
sns.set_palette("husl")

plt.figure(figsize=(12, 8))

contour_std = plt.contourf(Xg, Yg, Z_std, levels=50, cmap='Reds', alpha=0.5)

contour_p0 = plt.contourf(Xg, Yg, Z_p0, levels=50, cmap='Blues', alpha=0.5)

cbar1 = plt.colorbar(contour_p0, label='Density of $P_0$')
cbar1.ax.tick_params(labelsize=10)
cbar2 = plt.colorbar(contour_std, label='Density of Standard Normal')
cbar2.ax.tick_params(labelsize=10)

plt.scatter(X_t[0, :, 0], X_t[0, :, 1], 
            label="Initial samples $X_t[0]$", 
            c='blue', edgecolor='black', s=50, alpha=0.7)
plt.scatter(X_t[-1, :, 0], X_t[-1, :, 1], 
            label="Final samples $X_t[T-1]$", 
            c='orange', edgecolor='black', s=50, alpha=0.7)

plt.xlim(-3, 3)
plt.ylim(-2, 4)
plt.axis('equal')

plt.title("Distribution $P_0$ and Standard Normal with Samples", fontsize=14, fontweight='bold')
plt.xlabel("$x_1$", fontsize=12)
plt.ylabel("$x_2$", fontsize=12)
plt.legend(loc='upper right', fontsize=10, framealpha=0.9)

plt.tight_layout()
plt.show()
```

Сравните распределения вектора $X_t$ для двух видов обратных уравнений &mdash; стохастического и обыкновенного. Проведите сравнение для нескольких различных $t$.

```python
t_to_show = [int(i) for i in np.linspace(0, T-1, 10)]
```

Начнем с обратного стохастического

```python
X_restored_stochastic = np.zeros_like(X_t)
X_restored_stochastic[-1] = X_t[-1]

std_norm = sps.multivariate_normal(mean=[0, 0], cov=np.eye(2))
dt = 1.0

for i in range(T-1, 0, -1):
    A = np.sqrt(alpha_hat_t[i]) * Sigma + (1 - alpha_hat_t[i]) * np.eye(2)
    A_inv = np.linalg.inv(A)
    
    diff = X_restored_stochastic[i] - a * np.sqrt(alpha_hat_t[i])
    drift = -0.5 * beta_t[i] * X_restored_stochastic[i] + (beta_t[i] * (A_inv @ diff.T).T)
    
    epsilon = std_norm.rvs(size=X_restored_stochastic[i].shape[0])
    diffusion = np.sqrt(beta_t[i] * dt) * epsilon
    
    X_restored_stochastic[i-1] = X_restored_stochastic[i] - drift * dt - diffusion
```

Теперь обратное обыкновенное

```python
X_restored_ordinary = np.zeros_like(X_t)
X_restored_ordinary[-1] = X_t[-1]

dt = 1.0

for i in range(T - 1, 0, -1):
    A = np.sqrt(alpha_hat_t[i]) * Sigma + (1 - alpha_hat_t[i]) * np.eye(2)
    A_inv = np.linalg.inv(A)

    diff = X_restored_ordinary[i] - a * np.sqrt(alpha_hat_t[i])
    drift = -0.5 * beta_t[i] * X_restored_ordinary[i] + 0.5 * (
        beta_t[i] * (A_inv @ diff.T).T
    )

    X_restored_ordinary[i - 1] = X_restored_ordinary[i] - drift * dt
```

**3.** Делать перевод распределений из нормального в нормальное не так интересно, рассмотрим более сложный пример :)

Рассмотрим в качестве целевого распределения $\mathsf{P}_0$ равномерное на единичном кубе $[0, 1]^d$. Выпишите дифференциальные уравнения аналогично пункту 1.

*Указания.*
1. В данном случае на протяжении всего процесса координаты $X_t$ независимы, поэтому задачу достаточно решить 
для случая $d=1$.
2. Для нахожения распределения $X_t$ воспользуйтесь формулой свертки, связывающей распределение $X_0$ и условное $X_t|X_0$, ей мы воспользовались на лекции при оценке скор-функции. Сведите эту плотность к разности значений функции распределения стандартного нормального в двух точках. Продифференцировать ее вы легко сможете.

Начальное распределение:
$$
X_0 \sim U([0, 1]^d)
$$
$$
X_T \sim N(0, I_d)
$$

Как и сказано в указании, так как координаты $X_t$ независимы, то мы можем ограничиться случаем d=1.

В №1 мы получили 
$$
X_t|X_0 \sim N(\sqrt{\tilde{\alpha_t}}X_0, (1 - \tilde{\alpha}_t) I_d)
$$

\begin{align*}
p_t(x) &= \int_{0}^{1} p_0(x_0) \cdot q_t(x|x_0)dx_0 = \int_{0}^{1} N\left(x|\sqrt{\tilde{\alpha_t}}x_0, (1 - \tilde{\alpha}_t)\right) dx_0 \\
&= \int_{0}^{1}(2\pi (1 - \tilde{\alpha}_t))^{-1/2} \exp\left(-\frac{1}{2(1 - \tilde{\alpha}_t)}\left(x - \sqrt{\tilde{\alpha_t}}x_0\right)^2 \right)dx_0 \\
&=\left[a = \sqrt{\tilde{\alpha}_t}, \sigma = \sqrt{1 - \tilde{\alpha}_t}, y=ax_0, dx_0 =\frac{dy}{a} \right] \\
&= \frac{1}{a}\int_{0}^{a} \frac{1}{\sqrt{2\pi}\sigma}\exp\left(-\frac{1}{2\sigma^2}(x-y)^2 \right)dy \\
&= \left[z = \frac{x-y}{\sigma} \right] = \frac{1}{a}\int_{(x-a)/\sigma}^{x/\sigma}N(z|0, 1)dz =\frac{1}{a}\left(\Phi\left(\frac{x}{\sigma}\right) - \Phi\left(\frac{x-a}{\sigma}\right) \right)
\end{align*}

А как мы помним, производная функции распределения по аргументу явялется плотностью, тогда:
$$
\frac{\partial p_t(x)}{\partial x} = \frac{1}{a\sigma}\left(\phi\left(\frac{x}{\sigma}\right) - \phi\left(\frac{x-a}{\sigma}\right) \right)
$$

И сразу получаем производную логарифма:
\begin{align*}
\frac{\partial \log p_t(x)}{\partial x} = \frac{p_t'(x)}{p_t(x)} = \frac{1}{\sigma}\frac{\phi\big(\tfrac{x}{\sigma}\big)-\phi\big(\tfrac{x-a}{\sigma}\big)}
{\Phi\big(\tfrac{x}{\sigma}\big)-\Phi\big(\tfrac{x-a}{\sigma}\big)}
\end{align*}

Теперь выпишем дифференциальные уравнения, в них изменится только часть с производной логарифма.

Обратное стохастическое:
$$
dX_t = \left(-\frac{1}{2}\beta_t X_t - \beta_t \frac{1}{\sigma}\frac{\phi\big(\tfrac{x}{\sigma}\big)-\phi\big(\tfrac{x-a}{\sigma}\big)}
{\Phi\big(\tfrac{x}{\sigma}\big)-\Phi\big(\tfrac{x-a}{\sigma}\big)} \right)dt + \sqrt{\beta_t}dW_t
$$

Обратное обыкновенное:
$$
dX_t = \left(-\frac{1}{2}\beta_t X_t - \frac{1}{2}\beta_t \frac{1}{\sigma}\frac{\phi\big(\tfrac{x}{\sigma}\big)-\phi\big(\tfrac{x-a}{\sigma}\big)}
{\Phi\big(\tfrac{x}{\sigma}\big)-\Phi\big(\tfrac{x-a}{\sigma}\big)} \right)dt
$$

Прямое стохастическое то же, начальное распределения указаны в начале.

**4.** Повторите исследование из пункта 2 для дифференциальных уравнений из пункта 3.

Я возьму d=2 для лучшей визуализации.

```python
P_0 = sps.uniform(loc=0, scale=1)
T = 1000
t_s = np.arange(0, T)
beta_t = np.linspace(1e-2, 1e-4, T)
alpha_t = 1 - beta_t
alpha_hat_t = np.cumprod(alpha_t)
n_dots = 100
d = 2

X_0 = np.random.uniform(0, 1, size=(n_dots, d))

for i in range(t_s.shape[0]):
    mean = np.sqrt(alpha_hat_t[i]) * X_0
    cov  = (1 - alpha_hat_t[i]) * np.eye(d)
    X_t[i] = mean + np.random.randn(n_dots, d) * np.sqrt(1 - alpha_hat_t[i])
```

```python
def score_func(X_t, alpha_hat, t):
    a = np.sqrt(alpha_hat[t])
    sigma = np.sqrt(1 - alpha_hat[t]) + 1e-9
    u1 = X_t / sigma
    u2 = (X_t - a) / sigma
    numerator = sps.norm.pdf(u1) - sps.norm.pdf(u2)
    denominator = sps.norm.cdf(u1) - sps.norm.cdf(u2) + 1e-9
    score = (1 / sigma) * numerator / denominator
    return score
```

Снова начнем со стохастического

```python
X_restored_stochastic = np.zeros_like(X_t)
X_restored_stochastic[-1] = X_t[-1]

std_norm = sps.multivariate_normal(mean=[0, 0], cov=np.eye(2))
dt = 1.0

for i in range(T-1, 0, -1):
    drift = -0.5 * beta_t[i] * X_restored_stochastic[i] - beta_t[i] * score_func(X_restored_stochastic[i], alpha_hat_t, i)
    
    epsilon = std_norm.rvs(size=X_restored_stochastic[i].shape[0])
    diffusion = np.sqrt(beta_t[i] * dt) * epsilon
    
    X_restored_stochastic[i-1] = X_restored_stochastic[i] - drift * dt - diffusion
```

Теперь обыкновенное

```python
X_restored_ordinary = np.zeros_like(X_t)
X_restored_ordinary[-1] = X_t[-1]

dt = 1.0

for i in range(T - 1, 0, -1):
    drift = -0.5 * beta_t[i] * X_restored_ordinary[i] - 0.5 * beta_t[i] * score_func(X_restored_ordinary[i], alpha_hat_t, i)
    
    X_restored_ordinary[i - 1] = X_restored_ordinary[i] - drift * dt
```
