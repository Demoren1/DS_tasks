**1.** Рассмотрим в качестве целевого распределения $\mathsf{P}_0$ многомерное нормальное $\mathcal{N}(a, \Sigma)$. Найдите распределение зашумленного объекта на шаге $t$, используя явную формулу, а также производную его плотности по аргументу.

Мы уже вывели нужные формулы в прошлых номерах, просто применим их.

$X_t = \sqrt{\tilde{\alpha}_t} X_0 + \sqrt{(1 - \tilde{\alpha}_t)}\varepsilon_t \sim N(\sqrt{\tilde{\alpha}_t}a,\tilde{\alpha}_t\Sigma + (1 - \tilde{\alpha}_t)I_d)$

Обозначим дисперсию за A.

Тогда имеем следующую плотность

$$
p_t(x) = (2\pi \det A)^{-d/2}\exp\left(-\frac{1}{2}(x - \sqrt{\tilde{\alpha}_t}a)^T A^{-1}(x - \sqrt{\tilde{\alpha}_t}a)\right)
$$

Продифференцируем, получим:

$$
\nabla_{x}p_{t}(x) = -p_{t}(x)A^{-1}(x - \sqrt{\tilde{\alpha_t}}a)
$$

Заодно получим сразу градиент логарифма, он понадобится нам далее:
$$
\nabla_{x}\log(p_{t}(x)) = -A^{-1}(x - \sqrt{\tilde{\alpha_t}}a)
$$

Выпишите дифференциальные уравнения:
* прямое стохастическое,
* обратное стохастическое,
* обратное обыкновенное.

Укажите также начальное распределение.

Прямое стохастическое:

$$
dX_t = -\frac{1}{2}\beta_t X_t dt + \sqrt{\beta_t}dW_t,
$$
где $W_t$ -- итерация винеровского процесса.

Обратное стохастическое:
$$
dX_t = \left(-\frac{1}{2}\beta_t X_t +\beta_t A^{-1}(X_t - \sqrt{\tilde{\alpha_t}}a) \right)dt + \sqrt{\beta_t}dW_t
$$

Обратное обыкновенное:
$$
dX_t = \left(-\frac{1}{2}\beta_t X_t + \frac{1}{2}\beta_t A^{-1}(X_t - \sqrt{\tilde{\alpha_t}}a)\right)dt
$$

Начальное распределение:
$$
X_0 \sim N(a, \Sigma)
$$
$$
X_T \sim N(0, I_d)
$$

```python
import numpy as np
import scipy.stats as sps
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
a = np.array([2.0, 2.0])
Sigma = np.array([[2.0, 0.5], [0.5, 1.0]])
Sigma_inv = np.linalg.inv(Sigma)

P_0 = sps.multivariate_normal(mean=a, cov=Sigma)
```

```python
T = 1000
t_s = np.arange(0, T)
beta_t = np.linspace(1e-2, 1e-4, T)
alpha_t = 1 - beta_t
alpha_hat_t = np.cumprod(alpha_t)
```

```python
n_dots = 100
d = 2

means_t = np.sqrt(alpha_hat_t)[:, None] * a
covs_t = (
    alpha_hat_t[:, None, None] * Sigma[None]
    + (1 - alpha_hat_t)[:, None, None] * np.eye(d)
)

X_t = np.zeros((t_s.shape[0], n_dots, d))

for i in range(t_s.shape[0]):
    X_t[i] = sps.multivariate_normal(
        mean=means_t[i],
        cov=covs_t[i],
    ).rvs(size=n_dots)
```

```python
x = np.linspace(-4, 5, 500)
y = np.linspace(-3, 5, 500)
Xg, Yg = np.meshgrid(x, y)
pos = np.dstack((Xg, Yg))

Z_p0 = P_0.pdf(pos)
Z_std = sps.multivariate_normal(mean=[0, 0], cov=np.eye(2)).pdf(pos)

sns.set_style("whitegrid")
sns.set_palette("husl")

plt.figure(figsize=(12, 8))

contour_std = plt.contourf(Xg, Yg, Z_std, levels=50, cmap='Reds', alpha=0.5)

contour_p0 = plt.contourf(Xg, Yg, Z_p0, levels=50, cmap='Blues', alpha=0.5)

cbar1 = plt.colorbar(contour_p0, label='Density of $P_0$')
cbar1.ax.tick_params(labelsize=10)
cbar2 = plt.colorbar(contour_std, label='Density of Standard Normal')
cbar2.ax.tick_params(labelsize=10)

plt.scatter(X_t[0, :, 0], X_t[0, :, 1], 
            label="Initial samples $X_t[0]$", 
            c='blue', edgecolor='black', s=50, alpha=0.7)
plt.scatter(X_t[-1, :, 0], X_t[-1, :, 1], 
            label="Final samples $X_t[T-1]$", 
            c='orange', edgecolor='black', s=50, alpha=0.7)

plt.xlim(-3, 3)
plt.ylim(-2, 4)
plt.axis('equal')

plt.title("Distribution $P_0$ and Standard Normal with Samples", fontsize=14, fontweight='bold')
plt.xlabel("$x_1$", fontsize=12)
plt.ylabel("$x_2$", fontsize=12)
plt.legend(loc='upper right', fontsize=10, framealpha=0.9)

plt.tight_layout()
plt.show()
```

Сравните распределения вектора $X_t$ для двух видов обратных уравнений &mdash; стохастического и обыкновенного. Проведите сравнение для нескольких различных $t$.

```python
t_to_show = [int(i) for i in np.linspace(0, T-1, 10)]
```

Начнем с обратного стохастического

```python
X_restored_stochastic = np.zeros_like(X_t)
X_restored_stochastic[-1] = X_t[-1]

std_norm = sps.multivariate_normal(mean=[0, 0], cov=np.eye(2))
dt = 1.0

for i in range(T-1, 0, -1):
    A = np.sqrt(alpha_hat_t[i]) * Sigma + (1 - alpha_hat_t[i]) * np.eye(2)
    A_inv = np.linalg.inv(A)
    
    diff = X_restored_stochastic[i] - a * np.sqrt(alpha_hat_t[i])
    drift = -0.5 * beta_t[i] * X_restored_stochastic[i] + (beta_t[i] * (A_inv @ diff.T).T)
    
    epsilon = std_norm.rvs(size=X_restored_stochastic[i].shape[0])
    diffusion = np.sqrt(beta_t[i] * dt) * epsilon
    
    X_restored_stochastic[i-1] = X_restored_stochastic[i] - drift * dt - diffusion
```

Теперь обратное обыкновенное

```python
X_restored_ordinary = np.zeros_like(X_t)
X_restored_ordinary[-1] = X_t[-1]

dt = 1.0

for i in range(T - 1, 0, -1):
    A = np.sqrt(alpha_hat_t[i]) * Sigma + (1 - alpha_hat_t[i]) * np.eye(2)
    A_inv = np.linalg.inv(A)

    diff = X_restored_ordinary[i] - a * np.sqrt(alpha_hat_t[i])
    drift = -0.5 * beta_t[i] * X_restored_ordinary[i] + 0.5 * (
        beta_t[i] * (A_inv @ diff.T).T
    )

    X_restored_ordinary[i - 1] = X_restored_ordinary[i] - drift * dt
```

**3.** Делать перевод распределений из нормального в нормальное не так интересно, рассмотрим более сложный пример :)

Рассмотрим в качестве целевого распределения $\mathsf{P}_0$ равномерное на единичном кубе $[0, 1]^d$. Выпишите дифференциальные уравнения аналогично пункту 1.

*Указания.*
1. В данном случае на протяжении всего процесса координаты $X_t$ независимы, поэтому задачу достаточно решить 
для случая $d=1$.
2. Для нахожения распределения $X_t$ воспользуйтесь формулой свертки, связывающей распределение $X_0$ и условное $X_t|X_0$, ей мы воспользовались на лекции при оценке скор-функции. Сведите эту плотность к разности значений функции распределения стандартного нормального в двух точках. Продифференцировать ее вы легко сможете.

Начальное распределение:
$$
X_0 \sim U([0, 1]^d)
$$
$$
X_T \sim N(0, I_d)
$$

Как и сказано в указании, так как координаты $X_t$ независимы, то мы можем ограничиться случаем d=1.

В №1 мы получили 
$$
X_t|X_0 \sim N(\sqrt{\tilde{\alpha_t}}X_0, (1 - \tilde{\alpha}_t) I_d)
$$

\begin{align*}
p_t(x) &= \int_{0}^{1} p_0(x_0) \cdot q_t(x|x_0)dx_0 = \int_{0}^{1} N\left(x|\sqrt{\tilde{\alpha_t}}x_0, (1 - \tilde{\alpha}_t)\right) dx_0 \\
&= \int_{0}^{1}(2\pi (1 - \tilde{\alpha}_t))^{-1/2} \exp\left(-\frac{1}{2(1 - \tilde{\alpha}_t)}\left(x - \sqrt{\tilde{\alpha_t}}x_0\right)^2 \right)dx_0 \\
&=\left[a = \sqrt{\tilde{\alpha}_t}, \sigma = \sqrt{1 - \tilde{\alpha}_t}, y=ax_0, dx_0 =\frac{dy}{a} \right] \\
&= \frac{1}{a}\int_{0}^{a} \frac{1}{\sqrt{2\pi}\sigma}\exp\left(-\frac{1}{2\sigma^2}(x-y)^2 \right)dy \\
&= \left[z = \frac{x-y}{\sigma} \right] = \frac{1}{a}\int_{(x-a)/\sigma}^{x/\sigma}N(z|0, 1)dz =\frac{1}{a}\left(\Phi\left(\frac{x}{\sigma}\right) - \Phi\left(\frac{x-a}{\sigma}\right) \right)
\end{align*}

А как мы помним, производная функции распределения по аргументу явялется плотностью, тогда:
$$
\frac{\partial p_t(x)}{\partial x} = \frac{1}{a\sigma}\left(\phi\left(\frac{x}{\sigma}\right) - \phi\left(\frac{x-a}{\sigma}\right) \right)
$$

И сразу получаем производную логарифма:
\begin{align*}
\frac{\partial \log p_t(x)}{\partial x} = \frac{p_t'(x)}{p_t(x)} = \frac{1}{\sigma}\frac{\phi\big(\tfrac{x}{\sigma}\big)-\phi\big(\tfrac{x-a}{\sigma}\big)}
{\Phi\big(\tfrac{x}{\sigma}\big)-\Phi\big(\tfrac{x-a}{\sigma}\big)}
\end{align*}

Теперь выпишем дифференциальные уравнения, в них изменится только часть с производной логарифма.

Обратное стохастическое:
$$
dX_t = \left(-\frac{1}{2}\beta_t X_t - \beta_t \frac{1}{\sigma}\frac{\phi\big(\tfrac{x}{\sigma}\big)-\phi\big(\tfrac{x-a}{\sigma}\big)}
{\Phi\big(\tfrac{x}{\sigma}\big)-\Phi\big(\tfrac{x-a}{\sigma}\big)} \right)dt + \sqrt{\beta_t}dW_t
$$

Обратное обыкновенное:
$$
dX_t = \left(-\frac{1}{2}\beta_t X_t - \frac{1}{2}\beta_t \frac{1}{\sigma}\frac{\phi\big(\tfrac{x}{\sigma}\big)-\phi\big(\tfrac{x-a}{\sigma}\big)}
{\Phi\big(\tfrac{x}{\sigma}\big)-\Phi\big(\tfrac{x-a}{\sigma}\big)} \right)dt
$$

Прямое стохастическое то же, начальное распределения указаны в начале.

**4.** Повторите исследование из пункта 2 для дифференциальных уравнений из пункта 3.

Я возьму d=2 для лучшей визуализации.

```python
P_0 = sps.uniform(loc=0, scale=1)
T = 1000
t_s = np.arange(0, T)
beta_t = np.linspace(1e-2, 1e-4, T)
alpha_t = 1 - beta_t
alpha_hat_t = np.cumprod(alpha_t)
n_dots = 100
d = 2

X_0 = np.random.uniform(0, 1, size=(n_dots, d))

for i in range(t_s.shape[0]):
    mean = np.sqrt(alpha_hat_t[i]) * X_0
    cov  = (1 - alpha_hat_t[i]) * np.eye(d)
    X_t[i] = mean + np.random.randn(n_dots, d) * np.sqrt(1 - alpha_hat_t[i])
```

```python
def score_func(X_t, alpha_hat, t):
    a = np.sqrt(alpha_hat[t])
    sigma = np.sqrt(1 - alpha_hat[t]) + 1e-9
    u1 = X_t / sigma
    u2 = (X_t - a) / sigma
    numerator = sps.norm.pdf(u1) - sps.norm.pdf(u2)
    denominator = sps.norm.cdf(u1) - sps.norm.cdf(u2) + 1e-9
    score = (1 / sigma) * numerator / denominator
    return score
```

Снова начнем со стохастического

```python
X_restored_stochastic = np.zeros_like(X_t)
X_restored_stochastic[-1] = X_t[-1]

std_norm = sps.multivariate_normal(mean=[0, 0], cov=np.eye(2))
dt = 1.0

for i in range(T-1, 0, -1):
    drift = -0.5 * beta_t[i] * X_restored_stochastic[i] - beta_t[i] * score_func(X_restored_stochastic[i], alpha_hat_t, i)
    
    epsilon = std_norm.rvs(size=X_restored_stochastic[i].shape[0])
    diffusion = np.sqrt(beta_t[i] * dt) * epsilon
    
    X_restored_stochastic[i-1] = X_restored_stochastic[i] - drift * dt - diffusion
```

```python
def normalize(image):
    """
    Нормализация для изображений из MNIST
    """
    return (image * 2 - 1).to(torch.float32)

def unnormalize(image):
    return ((image.clamp(-1, 1) + 1) / 2 * 255).to(torch.uint8)

transform = v2.Compose([
    v2.ToImage(),
    v2.ToDtype(torch.float32, scale=True),
    v2.Lambda(normalize),
])

mnist_train = MNIST(root='.', train=True, transform=transform, download=True)
mnist_test = MNIST(root='.', train=False, transform=transform, download=True)
mnist = ConcatDataset([mnist_train, mnist_test])

def noise_images(images, steps, alpha_hat,device="cpu"):
    sqrt_alpha_hat = torch.sqrt(alpha_hat[steps])[:, None, None, None].to(device)
    sqrt_one_minus_alpha_hat = torch.sqrt(1 - alpha_hat[steps])[:, None, None, None].to(device)
    noise = torch.randn_like(images)
    return sqrt_alpha_hat * images + sqrt_one_minus_alpha_hat * noise, noise

def sample_images_one_step(
    noised_images,
    predicted_noise,
    steps,
    beta,
    alpha,
    alpha_hat,
    total_num_steps,
    add_noise=False,
):
    beta_step = beta[steps][:, None, None, None]
    if add_noise:
        noise = torch.randn_like(noised_images)
    else:
        noise = torch.zeros_like(noised_images)
    alpha_step = alpha[steps][:, None, None, None].to(device)
    alpha_hat_step = alpha_hat[steps][:, None, None, None].to(device)
    beta_step = beta[steps][:, None, None, None].to(device)
    return (
        1
        / torch.sqrt(alpha_step)
        * (
            noised_images
            - (1 - alpha_step) / (torch.sqrt(1 - alpha_hat_step)) * predicted_noise
        )
        + torch.sqrt(beta_step) * noise
    )


def sample_images(
    model,
    num_images,
    beta,
    alpha,
    alpha_hat,
    img_size=28,
    start_step=1000,
    finish_step=1,
    noised_images=None,
    device="cpu",
    return_full_trace=False,
):
    if noised_images is None:
        noised_images = torch.randn((num_images, 1, img_size, img_size)).to(device)

    full_traced = []
    for i in tqdm(
        reversed(range(finish_step, start_step)),
        position=0,
        total=start_step - finish_step,
    ):
        steps = (torch.ones(num_images) * i).long()
        with torch.no_grad():
            predicted_noise = model(noised_images, steps.to(device)).sample
        add_noise = True if i > 1 else False
        noised_images = sample_images_one_step(
            noised_images,
            predicted_noise,
            steps,
            beta,
            alpha,
            alpha_hat,
            start_step,
            add_noise=add_noise,
        )
        if return_full_trace:
            full_traced.append(noised_images.clone().detach().cpu())
    if return_full_trace:
        return noised_images, full_traced
    return noised_images

@torch.no_grad()
def visualize_training_progress(
    model,
    noise_scheduler,
    losses,
    total_num_steps: int,
    device="cpu",
    num_images: int = 9,
    nrow: int = 3,
):
    model.eval()

    # === график лоссов ===
    fig, (ax_loss, ax_imgs) = plt.subplots(1, 2, figsize=(14, 6))
    ax_loss.plot(np.arange(len(losses)) * 10, losses, color="teal")
    ax_loss.set_title("Loss history")
    ax_loss.set_xlabel("Iteration")
    ax_loss.set_ylabel("MSE loss")

    # === генерация изображений ===
    shape = (1, 28, 28)
    noisy = torch.randn((num_images, *shape), device=device)

    num_vis_steps = total_num_steps
    for t in reversed(range(num_vis_steps)):
        steps = torch.ones(num_images, device=device, dtype=torch.long) * t
        pred_noise = model(noisy, steps).sample
        noisy = noise_scheduler.step(pred_noise, t, noisy).prev_sample

    imgs = (noisy.clamp(-1, 1) + 1) / 2
    grid = make_grid(imgs, nrow=nrow)

    ax_imgs.imshow(grid.permute(1, 2, 0).cpu().numpy())
    ax_imgs.axis("off")
    ax_imgs.set_title("Generated samples")

    clear_output(wait=True)
    display(fig)
    plt.close(fig)
```

```python
def normalize_images(x):
    """Надёжная нормализация к [0, 1]"""
    if x.min() >= 0 and x.max() <= 1:
        return x.clamp(0, 1)
    elif x.min() >= -1 and x.max() <= 1:
        return ((x + 1) / 2).clamp(0, 1)
    elif x.max() > 1:
        return (x / 255.0).clamp(0, 1)
    return x.clamp(0, 1)


@torch.no_grad()
def compute_fid_score(real_images, fake_images, device="cpu", batch_size=64):
    """Вычисление FID между real_images и fake_images"""
    real_images = normalize_images(real_images).float()
    fake_images = normalize_images(fake_images).float()

    # повторяем каналы, если изображение одноканальное
    if real_images.size(1) == 1:
        real_images = real_images.repeat_interleave(3, dim=1)
        fake_images = fake_images.repeat_interleave(3, dim=1)

    fid_metric = FrechetInceptionDistance(feature=2048, normalize=True).to(device)

    for start in range(0, real_images.size(0), batch_size):
        r = real_images[start:start + batch_size]
        f = fake_images[start:start + batch_size]

        r = F.interpolate(r, size=(299, 299), mode="bilinear", align_corners=False)
        f = F.interpolate(f, size=(299, 299), mode="bilinear", align_corners=False)

        fid_metric.update(r.to(device), real=True)
        fid_metric.update(f.to(device), real=False)

    return fid_metric.compute().item()


@torch.no_grad()
def evaluate_fid(
    model, dataloader, betas_dict, device="cuda", num_images_to_sample=5000
):
    """Упрощённая версия: одна модель, один FID"""
    beta, alpha, alpha_hat = betas_dict

    # собираем изображения из даталоадера
    collected_real = []
    for real_imgs, _ in dataloader:
        collected_real.append(real_imgs)
        if len(collected_real) * real_imgs.shape[0] >= num_images_to_sample:
            break

    real_cat = torch.cat(collected_real, dim=0)[:num_images_to_sample].to(device)
    print(f"Диапазон реальных изображений: [{real_cat.min():.3f}, {real_cat.max():.3f}]")

    # сэмплируем из диффузии
    print(f"Сэмплирование {num_images_to_sample} изображений ...")
    fake_imgs = sample_images(
        model=model,
        num_images=num_images_to_sample,
        beta=beta,
        alpha=alpha,
        alpha_hat=alpha_hat,
        img_size=28,
        start_step=len(beta),
        finish_step=1,
        device=device,
    ).cpu()

    print(f"Диапазон сэмплированных изображений: [{fake_imgs.min():.3f}, {fake_imgs.max():.3f}]")

    # считаем FID
    fid_value = compute_fid_score(real_cat.cpu(), fake_imgs, device=device)
    print(f"FID = {fid_value:.2f}")

    return fid_value
```

```python
def init_training_modules(
    lr_warmup_steps, num_epochs, train_num_steps, beta_start, beta_end, device="cpu"
):
    """
    Инициализация различных модулей для обучения модели.
    """
    # Модель для восстановления шума
    model = UNet2DModel(
        sample_size=28,  # разрешение выходного изображения
        in_channels=1,  # количестов входных каналов
        out_channels=1,  # количество выходных каналов
        layers_per_block=2,  # количество ResNet слоев, используемых в каждом UNet блоке
        block_out_channels=(
            64,
            128,
            256,
        ),  # количество выходных каналов для UNet блоков
        down_block_types=(
            "DownBlock2D",  # стандартный ResNet downsampling блок
            "DownBlock2D",
            "DownBlock2D",
        ),
        up_block_types=(
            "UpBlock2D",  # стандартный ResNet upsampling блок
            "UpBlock2D",
            "UpBlock2D",
        ),
    )
    model.to(device)
    model.train()

    # Модуль контроля над уровнем шума -- взялли из Diffusers.
    # В данном случае он такой же, как в примере выше.
    noise_scheduler = DDPMScheduler(
        num_train_timesteps=train_num_steps, beta_start=beta_start, beta_end=beta_end
    )

    # Оптимизатор
    optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3)

    # Планировщик для скорости обучения (можно было обойтись без него)
    lr_scheduler = get_cosine_schedule_with_warmup(
        optimizer=optimizer,
        num_warmup_steps=lr_warmup_steps,
        num_training_steps=(len(dataloader) * num_epochs),
    )

    # Функция ошибки
    criterion = nn.MSELoss()

    output = (model, noise_scheduler, optimizer, lr_scheduler, criterion)

    return output
```

```python
def train_model_via_noise(
    model,
    dataloader,
    noise_scheduler,
    optimizer,
    lr_scheduler,
    criterion,
    num_epochs,
    total_num_steps,
    device="cpu",
    writer=None,
    plot_interval=200,
):
    losses = []
    global_iter = 0

    for epoch in range(num_epochs):
        for i, (images, _) in enumerate(tqdm(dataloader, "Training")):
            model.train()
            images = images.to(device)

            steps = torch.randint(0, total_num_steps, (images.size(0),), device=device)
            noise = torch.randn_like(images)
            noised_images = noise_scheduler.add_noise(images, noise, steps)
            predicted_noise = model(noised_images, steps).sample
            loss = criterion(noise, predicted_noise)

            if global_iter % 10 == 0:
                losses.append(loss.item())

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            lr_scheduler.step()
            global_iter += 1

            if (global_iter % plot_interval == 0) or (i == len(dataloader) - 1):
                visualize_training_progress(
                    model, noise_scheduler, losses, total_num_steps, device,
                )

    return losses
```

**2.** В формуле для ELBO мы получили среднюю квадратичную ошибку, домноженную на некоторый коэффициент, зависящий от $t$. На практике часто при обучении модели этот коэффициент отбрасывается.

Исследуйте, как меняется этот коэффициент в зависимости от $t$.

Помним, ELBO имеет следующий вид для репараметризованного:

\begin{align*}
ELBO = const - \sum_{t=2}^T \frac{\beta_t^2}{2\alpha_t \tilde{\beta}_t(1 - \tilde{\alpha}_t)}E_{q}||\delta_t - \delta_{\theta}(x_t, t)||^2.
\end{align*}

То же выражение без репараметризации:

\begin{align*}
ELBO = const - \sum_{t=2}^T \frac{\tilde{\alpha}_{t-1}\beta_t}{2\tilde{\beta}_t(1 - \tilde{\alpha}_t)^2}E_{q}||X_0 - X_{\theta}(X_t, t)||^2.
\end{align*}

```python
def calculate_factor_in_ELBO_noise(betas_dict, train_num_steps=1000):
    beta, alpha, alpha_hat = betas_dict[train_num_steps]

    beta = beta.detach().cpu().numpy()
    alpha = alpha.detach().cpu().numpy()
    alpha_hat = alpha_hat.detach().cpu().numpy()

    T = len(beta)
    beta_hat = (1 - alpha_hat[1:T - 1]) / (1 - alpha_hat[2:T]) * beta[2:]

    w_t = np.ones(T)
    w_t[2:T] = beta[2:] ** 2 / (2 * alpha[2:] * beta_hat * (1 - alpha_hat[2:]))
    return w_t

def calculate_factor_in_ELBO_image(betas_dict, train_num_steps=1000):
    beta, alpha, alpha_hat = betas_dict[train_num_steps]

    beta = beta.detach().cpu().numpy()
    alpha = alpha.detach().cpu().numpy()
    alpha_hat = alpha_hat.detach().cpu().numpy()

    T = len(beta)
    beta_hat = (1 - alpha_hat[1:T - 1]) / (1 - alpha_hat[2:T]) * beta[2:]

    w_t = np.ones(T)
    w_t[2:T] = alpha_hat[1:T-1] * beta[2:] / (2 * beta_hat * (1 - alpha_hat[2:])**2)
    return w_t
```

Для репараметризации

```python
def train_model_via_noise_with_coeff(
    model,
    dataloader,
    noise_scheduler,
    optimizer,
    lr_scheduler,
    criterion,
    num_epochs,
    total_num_steps,
    w_t,
    device="cpu",
    writer=None,
    plot_interval=200,
):
    losses = []
    global_iter = 0
    w_t = torch.as_tensor(w_t, dtype=torch.float32, device=device)

    for epoch in range(num_epochs):
        for i, (images, _) in enumerate(tqdm(dataloader, "Training")):
            model.train()
            images = images.to(device)

            steps = torch.randint(0, total_num_steps, (images.size(0),), device=device)
            noise = torch.randn_like(images)
            noised_images = noise_scheduler.add_noise(images, noise, steps)
            predicted_noise = model(noised_images, steps).sample
            
            loss_per_pixel = torch.square(noise - predicted_noise)  # [B, C, H, W]
            loss_per_sample = loss_per_pixel.mean(dim=(1, 2, 3))     # [B]
            weights = w_t[steps]                                     # [B]
            loss = (weights * loss_per_sample).mean()

            if global_iter % 10 == 0:
                losses.append(loss.item())

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            lr_scheduler.step()
            global_iter += 1

            if (global_iter % plot_interval == 0) or (i == len(dataloader) - 1):
                visualize_training_progress(
                    model, noise_scheduler, losses, total_num_steps, device,
                )

    return losses
```

Также исследуйте, как меняется квадратичная ошибка обученной модели в зависимости от $t$ при использовании коэффициента во время обучения и без его использования.

```python
def evaluate_mse_per_t(
    model,
    dataloader,
    noise_scheduler,
    device,
    T,
    predict_noise=True,
    plot_title="MSE по шагам времени t",
):
    model.eval()
    mse_per_t = torch.zeros(T, device=device)
    count_per_t = torch.zeros(T, device=device)

    with torch.no_grad():
        for (x0_batch, _) in tqdm(dataloader):
            x0_batch = x0_batch.to(device)
            batch_size = x0_batch.shape[0]
            t = torch.randint(0, T, (batch_size,), device=device)

            eps = torch.randn_like(x0_batch)
            alpha_bar_t = noise_scheduler.alphas_cumprod[t].view(-1, 1, 1, 1)
            xt = torch.sqrt(alpha_bar_t) * x0_batch + torch.sqrt(1 - alpha_bar_t) * eps

            pred = model(xt, t).sample

            if predict_noise:
                mse = (eps - pred).pow(2).mean(dim=[1, 2, 3])  # [B]
            else:
                mse = (x0_batch - pred).pow(2).mean(dim=[1, 2, 3])  # [B]

            for i in range(batch_size):
                ti = t[i].item() - 1
                mse_per_t[ti] += mse[i]
                count_per_t[ti] += 1

    mse_per_t = mse_per_t / (count_per_t + 1e-8)
    mse_per_t_np = mse_per_t.cpu().numpy()

    t_vals = np.arange(1, T + 1)

    plt.figure(figsize=(10, 6))
    plt.plot(
        t_vals,
        mse_per_t_np,
        marker="o",
        linestyle="-",
        linewidth=2,
        markersize=4,
        label="MSE(t)",
    )
    plt.xlabel("Шаг времени t")
    plt.ylabel("MSE(t)")
    plt.title(plot_title)
    plt.grid(True, alpha=0.3)
    plt.xticks(np.arange(0, T + 1, max(1, T // 10)))
    plt.tight_layout()

    plt.show()

    return mse_per_t_np
```

```python
mse_with_w_noise = evaluate_mse_per_t(
    model_via_noise_with_coeff,
    dataloader,
    noise_scheduler,
    device,
    T=train_num_steps,
    predict_noise=True,
    plot_title="MSE(t) — модель предсказывает шум, обучена с w_t"
)
```
